Class Imbalanced

```python
df['num_chars'] = df['message'].apply(len)
```

```python
df['num_words'] = df['message'].apply(lambda x: len(nltk.word_tokenize(x)))
```

```python
df['num_sentence'] = df['message'].apply(lambda x: len(nltk.sent_tokenize(x)))
```

```python
sns.pairplot(df, hue='label')
```

```python
df.describe()
```

```python
df[df['label'] == 0][['num_chars', 'num_words', 'num_sentence']].describe()
```

```python
df[df['label'] == 1][['num_chars', 'num_words', 'num_sentence']].describe()
```

- avg length of message in class **`Not Spam`** is arond **71**, while in **`Spam`** its around **138**.
- max. words in class **`Not Spam`** is arond **220**, while in **`Spam`** its around **46**, so not spam messages are generally longer than spam messages.
